## 1. Introduction

### 1.1. The Question / Estimand

Among completed best-of-three ATP singles matches between two right-handed players on outdoor hard courts in the 2022–2025 seasons, how much would a 10-percentage-point increase in Player A’s first-serve point conversion advantage change their probability of winning, after adjusting for the difference in pre-match ATP rankings?

For each match, we label the players A and B independently of who won. A player’s first-serve point conversion rate is the number of points they won on first serves divided by the number of first serves they landed. Player A’s advantage is their conversion rate minus Player B’s. The outcome is 1 if Player A won and 0 otherwise. We measure ranking advantage as the log of Player B’s pre-match ATP rank minus the log of Player A’s, so a positive value means Player A was ranked higher.

Our estimand is the average change in Player A’s win probability, measured in percentage points, when their first-serve conversion advantage increases by 10 percentage points. We hold each match’s pre-match ranking difference fixed and average over matches for which both advantage levels are supported by the data.

### 1.2. Data Description

We use four annual CSV files (`data/2022.csv` through `data/2025.csv`) from the [TennisMyLife match database](https://stats.tennismylife.org/tennis-match-database). Together they contain 11,932 match records before filtering. Each row represents one singles match and records the tournament and playing conditions, the players’ characteristics and rankings, the score, and serve statistics for the winner and loser.

We use `w_1stWon / w_1stIn` and `l_1stWon / l_1stIn` to calculate first-serve conversion for the winner and loser before mapping the rates to Players A and B. The fields `winner_rank` and `loser_rank` supply the pre-match rankings. We use `surface`, `indoor`, `best_of`, `winner_hand`, `loser_hand`, and `score` to identify completed best-of-three outdoor hard-court matches between right-handed players. The analysis requires recorded ranks and a positive count of landed first serves for both players.

The dataset records both players’ serve performance under the same match conditions, and their pre-match rankings let us adjust for a proxy for relative skill. Restricting surface, setting, match format, handedness, and era makes matches more comparable. A causal interpretation still requires the assumption that the ranking adjustment and sample restrictions account for the common causes of first-serve conversion advantage and winning. Factors such as form or injury may affect both.
